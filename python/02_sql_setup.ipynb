# SQL Setup — Load Data into SQLite
This notebook takes the `master.csv` file we created and loads it into a SQLite database.
SQLite is a simple database stored in a single file — no server needed.
Once the data is in the database, we can run SQL queries on it.

```python
import sqlite3
import pandas as pd

# load the master table we built in the previous notebook
master = pd.read_csv('../data/master.csv')

# connect to (or create) the SQLite database file
conn = sqlite3.connect('../sql/olist.db')

# write the main table — if it already exists, replace it
master.to_sql('sales', conn, if_exists='replace', index=False)

# also write smaller lookup tables for customers and products
master[['customer_unique_id', 'customer_city', 'customer_state']] \
    .drop_duplicates() \
    .to_sql('customers', conn, if_exists='replace', index=False)

master[['product_id', 'category_english']] \
    .drop_duplicates() \
    .to_sql('products', conn, if_exists='replace', index=False)

conn.close()
print('Done. Database saved to ../sql/olist.db')
```
